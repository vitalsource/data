```python
import pandas as pd
```

## Read dataset of student–AI tutor interactions

```python
interactions = pd.read_parquet( 'interactions.parquet' )
interactions.head()
```

## Phase 1: Cognitive Load Theory for AI Response Analysis

### Methods

>The final dataset contains 500 randomly sampled interactions across 170 textbooks and 305 students.

```python
len( interactions ), interactions.textbook_id.nunique(), interactions.student_id.nunique()
```

>Using the Book Industry Standards and Communications major subject heading classification, the top subject domains as a percentage of interactions were Social Science (30.8%), History (11.4%), Business & Economics (9.4%), Psychology (8.4%), and Political Science (8.0%).

```python
interactions.subject.value_counts( normalize=True ).apply( lambda p: f'{p:.1%}' )
```

### Results and Discussion

>The correctness judge classified the 500 interactions as 81.8% correct, 13.4% incomplete, and 4.8% incorrect.

```python
interactions.original_correctness.value_counts( normalize=True ).apply( lambda p: f'{p:.1%}' )
```

>The CLT judge identified that 29.6% of the tutor's responses contained extraneous information.

```python
interactions.original_ecl.value_counts( normalize=True ).apply( lambda p: f'{p:.1%}' )
```

>**Table 1.**  Distribution of extraneous cognitive load across tutor response correctness categories.

```python
pd.crosstab( interactions.original_correctness, interactions.original_ecl )
```

## Phase 2: Cognitive Load Theory as an AI Revision Agent

### Results and Discussion

>The same correctness analysis was performed on the revised responses, with incomplete responses reduced from 13.4% to 0.6% and incorrect responses reduced from 4.8% to 1.0%.

```python
interactions.revised_correctness.value_counts( normalize=True ).apply( lambda p: f'{p:.1%}' )
```

>The overall presence of extraneous content was reduced from 29.6% to 0.8%.

```python
interactions.revised_ecl.value_counts( normalize=True ).apply( lambda p: f'{p:.1%}' )
```

>**Table 2.**  Distribution of extraneous cognitive load across tutor response correctness categories after CLT-based revision.

```python
pd.crosstab( interactions.revised_correctness, interactions.revised_ecl )
```

>The CLT agent reduced the original AI tutor response from 184 words to 60 while still providing the necessary definitions and relationship between terms.

```python
# Behaviorism/positivism example from the paper
example = interactions[ interactions.original_response.str.contains( 'Behaviorism', na=False ) ].iloc[ 0 ]
original_words = len( example.original_response.split() )
revised_words  = len( example.revised_response.split() )
print( f'Original: {original_words} words' )
print( f'Revised:  {revised_words} words' )
print()
print( 'Original response:' )
print( example.original_response )
print()
print( 'Revised response:' )
print( example.revised_response )
```
